# Numpy Array Properties

```python
import numpy as np
array_1D = np.array([1,2,3])
array_2D = np.array([[1,2,3],[5,8,7]])
array_3D = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
```

1. Shape 
    * When working with multi-dimensional data , shape help to uderstand structure. 
    *  array.shape
    * Return (rows, columns)


```python
dim = array_2D.shape
print(dim)
```

2. Size
    * Total number of elements in array
    * array.size

```python
size = array_2D.size
print(size)
```

3. ndim
    * Return no. of dimension 
    * array.ndim

```python
print("1D array : ",array_1D.ndim)
print("2D array : ",array_2D.ndim)
print("3D array : ",array_3D.ndim)
```

4. Data Type
    * return data type of array elements
    * array.dtype

```python
print(array_1D.dtype)
```

5. Data type conversion
    * arr.astype(int)
    * Return's new array , not update arr

```python
arr = np.array([1.01,2.25,4.43,0.98])

print("Initial Data type of arr : ",arr.dtype)
int_arr = arr.astype(int)
print("int arr : ", int_arr," Data Type : ", int_arr.dtype)
str_arr = arr.astype(str)
print("str_arr : ", str_arr," Data Type : ", str_arr.dtype)
```
