# NumPy Operations On Array

* We can perform mathematical operation array without using loop.
* It perform operation very fast.

```python
import numpy as np 
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([9,8,7,6,5])
```

1. Mathematical Operators

```python
print("Adding 2 in each element in arr1 : ",arr1+2)
print("Subtrating 2 from each element in arr1 : ",arr1-2)
print("Multipling 2 with each element in arr1 : ",arr1*2)
print("Taking sq of each element of arr1 : ",arr1**2)
print("Dividing each element of arr2 by 2 : ",arr2/2)

# We can perform all opeartion on two array but their length should be equal
print("Addition of two array : ",arr1+arr2)
```

2. Aggregation Function
    * np.sum(arr) --> add all element of array.
    * np.mean(arr) --> give avg of array.
    * np.max(arr) --> min value of array.
    * np.min(arr) --> min value of array.
    * np.std(arr) --> standard deviation.
    * np.var(arr) --> cal var of array.


```python
print("Sum : ", np.sum(arr1))
print("Avg : ",np.mean(arr1))
print("Max : ", np.max(arr1))
print("Min : ",np.min(arr1))
print("STD : ",np.std(arr1))
print("Var : ",np.var(arr1))
```
