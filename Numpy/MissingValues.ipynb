# Handling Missing Values In NumPy

* Missing values are data values that are not available or unknown.
* np.nan can only be store in floating point arrays. 

```python
import numpy as np
arr = np.array([1,2,np.nan,4,6,np.nan])
```

1. np.isnan(array)
    * is nan -> not a number
    * return true at its index.

```python
print("Array : ", arr,"\n")
print("isnan : ", np.isnan(arr))

print("Count missing values : ",np.isnan(arr).sum())

# We can't compare nan values.
```

2. replacing nan values

```python
arr2 = np.array([1,np.nan,3])
arr2[np.isnan(arr2)] = 0
print(arr2)
```

3. np.nan_to_num(array, nan = value) 
    * Replace nan values with specific function.
    * default = 0

```python
cleaned_arr = np.nan_to_num(arr)
print("Cleaning nan values with default 0 :",cleaned_arr)
cleaned_arr2 = np.nan_to_num(arr,nan=1000)
print("Cleaning nan values with default 2 :",cleaned_arr2)
```

4. np.isinf(array)
    * Check wheather array contain infinity values.
    * Return boolean array.

```python
inf_arr = np.array([np.inf, 2 ,3 , 4, -np.inf,10])
print("Array containing infinity values : ",inf_arr,"\n")
print(" Checking wheather contains infinity values : ",np.isinf(inf_arr))
```

5. Replacing Infinity values.
    * np.nan_to_num(array , posinf=value, neginf=value)

```python
inf_cleaned = np.nan_to_num(inf_arr,posinf = 0, neginf =0)
print("Clean infinity values : ", inf_cleaned)
```

6. Aggregate Function on nan value array.
    * np.nanmean(arr)
    * np.nansum(arr)
    * np.nanmax(arr)
    * np.nanmin(arr)

```python
arr = np.array([1,100, np.nan, 67, 89, np.nan, 34])

print(np.isnan(arr))

print("Sum : ",np.nansum(arr))
print("Mean : ", np.nanmean(arr))
print("Max : ", np.nanmax(arr))
print("Min : ", np.nanmin(arr))
```
