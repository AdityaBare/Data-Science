# Reshaping And Manipulating Arrays

* Modifing dimensions (shape) of array witout changing data.
* Example :- convert 1D arr to 2D arr without changing data ( element size remains same).

```python
import numpy as np
arr = np.array([10,20,30,40,50,60,70,80,90])
arr2D = np.array([[1,2,3],[4,5,6]])
```

1. reshape( rows, columns) specify new shape
    * if dimension match (we can divide equaly in rows and columns)
    * Reshaping doesnot create copy , it return view.
    * Means Changing value in new array effict old array.

```python
reshape_array = arr.reshape(3,3)
print(reshape_array,"\n\n")

arr[0] = 3

print(arr,"\n\n")

print(reshape_array,"\n")

print("Convering into 1D array",arr2D.reshape(arr2D.size))

```

2. Flattening Array
    * .ravel()  -> It return a *view* (Direct convert multi-dimensional array into 1D array)
    * .flatten() -> It return *copy* (Direct convert multi-dimensional array into 1D array).
    * Both use to convert multi-dimensional array into 1D array.

```python
print("Original 2D array :", arr2D,"\n")
print("Ravel return view (new array can reflect value in old one) : ",arr2D.ravel())
print("Flatten return copy (new array can't refelect values in old one :)", arr2D.flatten())
```
