# Vectorization 

* Vectorization is the process of performing operations on an entire array at once instead of using loops( for or while).

* It makes NumPy programs:
    1. Faster
    2. Simpler
    3. More memory efficient

```python
import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([10,20,30])
```

1. Addition of two number.

```python
print(arr1+arr2)
```

2. Multiplication

```python
print(arr1*2) 
# Here broadcasting is also used.
```

3. np.sqrt(array)

```python
print(np.sqrt(np.array([1,4,9,16,36,49])))
```
