# Array Manipulation #

* NumPy Array Manipulation includes:
    * Reshaping (reshape(), resize())
    * Changing dimensions (flatten(), ravel(), transpose())
    * Adding elements (append(), insert())
    * Removing elements (delete())
    * Joining/Stacking (concatenate(), stack(), vstack(), hstack())
    * Splitting (split(), vsplit(), hsplit())

* Python List is store mixed datatype, numpy array store element of same datatype.
* Python list -> Dynamic .
* NumPy array -> Fixed Size Creation.
* As array is fixed size when we perform operation like remove , insert that time new array is created(or sreturn new array).




```python
import numpy as np
arr = np.array([1,2,3,4,5])
arr2 = np.array([5,6,7,8,9])
arr2D  = np.array([[1,2,3],[4,5,6]])
```

1. np.insert(array , index , element , asix=None)
    * array - original array.
    * index - position where insert data.
    * element - data to insert.
    * axis -None(default) ( if it is none data inserted into flatten array).
    *                     ( 0-> insert into row, 1-> insert into column).

```python
ins_array = np.insert(arr,2,225)
print("After inserting 225 at index 2 :", ins_array)

ins_arr2D = np.insert(arr2D,3, [225],axis=1)
print("After inserting 225 at index 3 ",ins_arr2D,"\n")

print(np.insert(arr2D , 2,[7,8,9], axis=0))

print("It is 2D array but make it flatten then add at give index :",np.insert(arr2D , 3, 35))
```

2. np.append(array , value , axis=None)
    * same as like insert .
    * but only add values at they end.

```python
append_arr = np.append(arr , [6,7,8,9])
print(append_arr)

print(np.append(arr, 78))

append_arr2D = np.append(arr2D , [[11,12,13]] , axis=0)
print(append_arr2D)
```

3. np.concatenate((array1, array2), axis = 0)
    * axis = 0 -> set row.
    * axis = 1 -> set columns.

```python
print(np.concatenate((arr,arr2), axis = 0))
```

4. np.delete(array, index , axis =None)

```python
delete_element_array = np.delete(arr,2)
print(delete_element_array,"\n")

delete_2D_Array = np.delete(arr2D,1, axis=1)
print(delete_2D_Array,"\n")
print( np.delete(arr2D,1, axis=0))
```

5. Stacking ( Merging data row or columns wise.)
    * np.vstack((arr1, arr2)) -> row wise vertical stack.
    * np.hstack((arr1, arr2)) -> columns wise horizontal stack.

```python
print("Horizontal Stacking : ", np.hstack((arr, arr2)))
print("Vertical Stacking : ",np.vstack((arr,arr2)))
```

6. Spliting 
    * Dividing array into sub arrays.
    * np.split(arr, parts) -> divide into equal parts. ( default axis =0 )
    * np.hsplit() -> split horizontal. (axis=0 , split only by rows.)
    * np.vsplit() -> vertical split.   (axis=1 , split only by columns.)

```python
divisible_array = np.array([1,2,3,4,5,6])

# Array should equally divisible by the part given in  split function , other wise function give error.
print("Dividing 6 element array into 3 parts :",np.split(divisible_array,3))
print("Dividing 6 element array into 3 parts :",np.split(divisible_array,2),"\n")

print(np.hsplit(divisible_array,2))
print(np.vsplit(arr2D,2))
```
