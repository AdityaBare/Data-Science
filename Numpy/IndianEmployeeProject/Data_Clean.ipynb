```python
import numpy as np
import pandas as pd
```

* Loading data in np

```python
data = np.genfromtxt("indian_employee_data.csv",delimiter=',',skip_header=1)
print(data.shape)
```

Loading data in pandas

```python
df = pd.read_csv("indian_employee_data.csv")
print(df.head()) #Return first fives rows of data set.
```

* Checking the missing values.

```python
print("Missing values in each columns \n")
print(df.isnull().sum(),'\n\n')

negative_count = (df.select_dtypes(include='number')<0).sum()
print(negative_count)
```

* Filling empty values.

```python
df['salary_inr'].fillna(df['salary_inr'].median() , inplace=True)
df['experience_yrs'].fillna(df['experience_yrs'].median() , inplace=True)
```

* Replacing infinit values 

```python
df.replace([np.inf, -np.inf],np.nan , inplace=True)
df.fillna(df.mean(), inplace=True)
```

* Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

* Replacing Negative values.

```python
df['salary_inr'] = np.where(df['salary_inr']<0, df['salary_inr'].mean(), df['salary_inr'])

df['experience_yrs'] = np.where(df['experience_yrs']<0, 0, df['experience_yrs'])

df['attendance_pct'] = np.where(df['attendance_pct']<0, 0, df['attendance_pct'])

df['age'] = np.where(df['age']<0, df['age'].mean(), df['age'])
```

* Removing salary then upper and lower bound.

```python
salary_mean = df['salary_inr'].mean()
salary_std = df['salary_inr'].std()
lower_bound = salary_mean-(3*salary_std)
upper_bound = salary_std+(3*salary_std)

df = df[(df['salary_inr']>= lower_bound) & (df['salary_inr']<= upper_bound)]
```

```python
df.to_csv('cleaned_indian_employee_data.csv' , index=False)

print("Data cleaning completed! Saved as 'cleaned_indian_employee_data.csv' ")
```
