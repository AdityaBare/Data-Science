# Level 7 — Time-Series Analysis

* Add Column 

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
```

```python
df.head()
```

* Total sales by year

```python
df.groupby('Year')['Sales'].sum()
```

* Total sales by month

```python
sales_per_month = df.groupby('Month Name')['Sales'].sum()
sales_per_month
```

* Which Month has highest sales

```python
sales_per_month.sort_values(ascending=False).nlargest(1)
```

* Which Month has lowest sales.

```python
sales_per_month.sort_values(ascending=True).nlargest(1)
```

* Calculate monthly average sales

```python
df.groupby('Month Name')['Sales'].mean()
```

* Find the best-performing year.

```python
df.groupby('Year')['Sales'].sum().sort_values(ascending=False).nlargest(1)
```
