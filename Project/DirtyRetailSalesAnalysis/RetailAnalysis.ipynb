# Invalid Numerical Data

1. Find Negative quantity

```python
df[df['Quantity']<0]
```

```python
# Sales = Quantity * price * (1-dis)


df['Quantity'] = np.where(
    df['Quantity']<0 , df['Sales']/(df['Price Each']*(1-df["Discount"])), df['Quantity']
)
```

* Find  invalid Price

```python
df[df['Price Each']<0]
```

```python
# Sales = Quantity * price * (1-dis)
#  price = (sales)/(Quantity *(1-dis))
df['Price Each'] = np.where(
    df['Price Each']<0, df['Sales']/(df['Quantity']*(1-df["Discount"])) , df['Price Each'])
```

3. Check Discount

```python
df['Discount'].describe()
```

```python
df[df['Discount']>1]
```

```python
df['Discount'] = np.where(
    df['Discount']>1 , df['Discount']*0.1 , df['Discount']
)
```
