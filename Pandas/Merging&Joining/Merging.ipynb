# Merging

- Merging DataFrames in Pandas is similar to performing SQL joins. It is useful when we need to combine two DataFrames based on a common column or index. The merge() function provides flexibility for different types of joins.
- There are four basic ways to handle the join (inner, left, right and outer) depending on which rows must retain their data.

```python
import pandas as pd

data1 = {
    'key':['id1','id2','id3','id4','id5'],
    'Name':['Sahil',"Aditya","Pranav","Nilesh",'Tejas'],
    "Age":[18,20,19,20,23]
}

data2 ={
    'key':['id4','id3','id1','id2','id6'],
    'Address':['Vinchur','Sambhajinager','Dhamori','Dhamori','Nagpur'],
    'Qualifications':['BTech',"Sports","12th",'BTech','MBA']
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

```

- Cross Join
    - Here we don't use 'on' 
    - each row of df1 combine with each row of df2

```python
read = pd.merge(df1, df2,  how ='cross')
print(read)
```

- read = pd.merge(df1 , df2 , how = 'inner/left/right/outer' on='id')
    1. pass DataFrame that gwt merge
    2. how = inner or left or right or outer join
    3. on = merge data with one unique key combination.
    - Here we can also pass multiple keys contain both data frame in form of list.
    4. If both column name is diffrent but contain same unique id's then simple use , example :- left='ID', right='UserID' in that function.

1. Inner Join
    - By default it follow's inner join

```python
read = pd.merge(df1, df2 , on='key')
print(read)
```

2. Left join

```python
read = pd.merge(df1, df2 , how='left' , on='key')
print(read)
```

2. Right Join

```python
read = pd.merge(df1, df2 , how='right', on='key')
print(read)
```

4. Outer Join

```python
read = pd.merge(df1, df2 , how='outer' , on='key')
print(read)
```
