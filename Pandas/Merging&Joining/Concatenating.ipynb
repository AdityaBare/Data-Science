# Concatenating DataFrames

- Concatenating DataFrames means combining them either by stacking them on top of each other (vertically) or placing them side by side (horizontally). In order to Concatenate dataframe, we use different methods which are as follows:

```python
import pandas as pd

data1 = {
    'Name':['Aditya','Sahil','Tejas'],
    'Age':[20,18,23]
}
data2 = {
    'Name':['Pranav',"Nilesh",'Tejas'],
    'Age':[19,20,23]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

1. Concatenating DataFrame using .concat()

```python
read =pd.concat([df1,df2]) 
read
```

```python
pd.concat([df1,df2], ignore_index=True) # We get correct indexing
```

2. Add data horizontally

```python
read =pd.concat([df1,df2], axis = 1, ignore_index=True)
read
```

3. Add Data vertically

```python
read =pd.concat([df1,df2], axis = 0, ignore_index=True)
read
```
