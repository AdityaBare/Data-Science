1. df.head()
    - .head() -> first five row.
    - .head(n) -> n number of rows.
    1. Check data is correctly load.
    2. How data is organize.          
    3. Clear understanding of data.

```python
print("First 5 row of Dataset.\n")
print(df.head(),'\n\n\n\n')
print("Display first 10 rows by parse in head\n")
print(df.head(5))
```

2. df.tail()
    - same as head by last rows.

```python
print("Last 5 rows\n")
print(df.tail(),"\n\n\n\n")
print("Display last 10 row by parse in tail\n")
print(df.tail())
```

3. df.info()
    - It is method which gives summary of your dataset.
    - It Gives
        1. Number of rows and columns.
        2. column name.
        3. data types .
        4. non null count.
        5. memory usage of data frame.

```python
print("Displaying the info of dataset.\n\n")
print(df.info())
```

4. df.describe()
    - Generates descriptive statistics for numeric columns in your dataframe.
    - Statistics is science of learning from data.
    - It Gives :
        1. Count :- Number of non missing values in column.
        2. Mean :- is the average of a set of numbers.
        3. Standard Deviation (std) : measures how spread out the values are from the mean (average).
        4. min : smallest value.
        5. 25% : (First Quarter) 25% of data is less than or equal to this data.
        6. 50% : Half (50%) of the data is below this value, and half is above it.
        7. 75% of the data is less than or equal to this value.

```python
print("Dataset description\n")
print(df.describe(),"\n\n")
```

```python
print("DataFrame description\n")
print(dfs.describe())
```
