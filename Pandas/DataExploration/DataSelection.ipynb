# Data Selection 

* Selecting specific column.
* Filter rows.
* Combine multiple conditions.

- Selecting columns -> use square brackets.
    1. Return a Series.
    2. OR return DataFrame for multiple columns of data.
- Filter row -> boolean conditions.

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\barea\OneDrive\Desktop\Data Science\Numpy\IndianEmployeeProject\indian_employee_data.csv")

data = {
    "Name":["Nilesh","Pranav","Aditya","Ram"],
    "Age":[20,19,20,20],
    'marks':[1,2,3,4]
}
dfs = pd.DataFrame(data)
```

1. Select column
    - column = df["col name"]

```python
subset = df[['salary_inr','age','rating']]
print("Selecting Multiple combination")
print('\n',subset,'\n')

Salary = df['salary_inr']
print("Only Age column")
print(Salary)
```

2. Filtering  
    - filter_row = df[df["Salary"]>50000] -> Single Condition.
    - filter_row = df[(df["salary"]>5000) & (df["Salary"]<80000>)]  -> For multi condition use ().

```python
# Filtering on single condition.
filter_age = df[df["age"]>df['age'].mean()]
print("Filter on one condition\n")
print(filter_age,"\n\n")
```

* Filtering with And(&)

```python
# Filtering on multiple condition
print("Condition on age and salary.\n")
print(df[(df['salary_inr']>300000) & (df['age']>34)])
```

* Filtering with OR(|)

```python
# Filtering on multiple condition
print("Condition on age or salary.\n")
print(df[(df['salary_inr']>300000) | (df['age']>34)])
```
