# Summary Statistics

* Aggregate Function 
     1. min()
     2. max()
     3. sum()
     etc
* df['col'].fun()


```python
import pandas as pd
import numpy as np

age = [12, 35, 87,23, np.nan]
df = pd.DataFrame(age , columns=['age'])
print(df)
```

```python
print("Min : ",df['age'].min())
print("Max : ",df['age'].max())
print("sum : ", df['age'].sum())
print("Count : ", df['age'].count())
print("Mean : ",df['age'].mean())
print("Median : ", df['age'].median())
```
