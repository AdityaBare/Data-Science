# Selecting On Group

```python
import pandas as pd 

data = {
    "Fruit" : ['apple','banana','mango','pineapple','apple'],
    'price' :[120,40,60,44,87],
    'Amount':[100,56,57,17,56]
    }
df = pd.DataFrame(data)
```

1. df.groupby("col_1")["col2"].agg()

    - df['Amount']          # select column
    - df.groupby('Fruit')   # call a method → ()
    - agg(['sum', 'mean', 'max', 'min'])

```python
group = df.groupby('Fruit')['Amount'].sum()
print(group)
```

```python
print(df.groupby("Fruit")['price'].mean())
```

2. Appling grupby on multiple columns

```python
group = df.groupby(['Fruit','Amount'])['price'].mean()
print(group)
```
