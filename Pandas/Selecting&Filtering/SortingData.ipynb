# Sorting Data

* Sorting helps to arrange data into meaningful order.

```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\barea\OneDrive\Desktop\Data Science\Numpy\IndianEmployeeProject\indian_employee_data.csv")
```

1. Sorting data 1 Column 
    - df.sort_values(by="Column Name" , ascending=True/False , inplace =  True)
    - by = pass column name.
    - True = Ascending order
    - False = Descending order.
    - inplace = True -> change original df.

```python
df.dropna(axis=0 , inplace = True)
df1 =df.sort_values(by='salary_inr',ascending=True , inplace=False)


print("Sorted Data Frame by ")
print(df1)
```

2. Sorting Data on multiple columns.
    - by = ['col1','col2']
    - ascending = [True , False]

```python
df.sort_values(by=['age','rating'] ,ascending=[False, True], inplace =True)
print(df)
```
