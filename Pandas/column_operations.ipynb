# Data Updation On Columns

* add column
* update column
* remove column

```python
import pandas as pd

data = {
    "Name":["Nilesh","Pranav","Aditya","Ram"],
    "Age":[20,19,20,20],
    'marks':[1,2,3,4],
    "Salary":[30000,25000,35000,10000]
}
df = pd.DataFrame(data)
print(df)
```

# Add columns

1. df["Column Name"] = some_Date (list , series, or caluculation )

```python
df["Bonus"] = df['Salary']*0.1
print(df)
```

2. using insert()
    - using insert method we can insert column at specific index.
    - df.insert(loc, "Col_Name", some_Data)

```python
df.insert(0,"Roll No.",[32,28,41,90])
print(df)
```

# Updating Values

1. Upadting whole column

```python
df['Salary'] = df["Salary"]*1.05
print(df)
```

2. df.loc[] 
    - Is a method in pandas through it we can access specific cell.
    - df.loc[row_index , "Column_Name"] = new_value

* Update specific cell

```python
df.loc[1,"Age"] = 20
print(df)
```

* Selcting Specific Row & Columns

```python
df.loc[[0,2],["Name","Roll No."]]
```

* Other use case of loc

```python
print("To print specific row \n")
print(df.loc[2],'\n\n')

print("Select only column\n")
print(df.loc[:,['Name',"Age"]],'\n\n')

print("To print column with filter on other column\n")
print(df.loc[df['Salary']>30000, 'Name'])
```

# Drop Column

1. Drop
    - df.drop(columns=["Column_Name],inplace=True)
    - inplace = True -> change original data frame.
    - inplace = False -> return new DataFrame.

```python
df.drop(columns=['marks','Bonus'], inplace=True)
print(df)
```
