# Saving Files After Manipulation.

```python
import pandas as pd
```

```python
data = {
    "Name":['Ram','Shyam','Gopal'],
    "Age":[10,20,30],
    "City":["Nagpur","Mumbai","Delhi"]
}
```

* Save data in tabular format.

```python
df = pd.DataFrame(data)
print(df)
```

* Save file into CSV format.
    - df.to_csv("FileName.csv") -> In this file it include also index per row.
    - df.to_csv("FileName.csv",index=False) -> In this it does not include index.

```python
df.to_csv("SampleSaveFile.csv", index=False)
```

* Saving files into Excel format 
    - with extension xlsx.
    - df.to_excel("FileName.xlsx",index=False)

```python
df.to_excel("SaveSampleFile.xlsx" , index=False)
```

* Save in json format

```python
df.to_json("SampleSaveFile.json" )
```
