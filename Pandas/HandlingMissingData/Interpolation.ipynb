# Interpolation
    - Interpolation means filling missing (NaN) values by estimating a   value between known values.
    - Why Interpolation ?
        1. Prevent data integrity.
        2. Smooth treads.
        3. Avoid data loss.

1. df.interpolate(method="method" , axis=0 , inplace = True)
    - Method 
        1. linear (y=mx+b) -> It assumes the values change at a constant rate.
        2. polynomial (y=ax2+bx+c) -> Polynomial interpolation uses a polynomial curve to estimate missing values.
        3. time -> Used when your DataFrame has a time/date index.
        4. nearest -> Uses the value from the nearest known point.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A':[1,2,np.nan,4],
    'B':[5,np.nan,np.nan,8],
    'C':[np.nan,10,11,12]
})
```

1. linear

```python
df.interpolate(method='linear', limit_direction='both') 

```
