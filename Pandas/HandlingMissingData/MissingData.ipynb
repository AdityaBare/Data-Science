```python
import pandas as pd

df2 = pd.read_csv(r"C:\Users\barea\OneDrive\Desktop\Data Science\Numpy\IndianEmployeeProject\indian_employee_data.csv")


data = {
    "Roll No.":[1,None,5,20],
    "Name":["Nilesh","Pranav","Aditya","Ram"],
    "Age":[20,19,None,20],
    'marks':[1,2,3,4],
    "Salary":[30000,25000,35000,None]
}
df = pd.DataFrame(data)
print(df)
```

1. isnull()
    - True = Value is misssing.
    - False = Value is present.

```python
df.isnull()
```

```python
df2.isnull().sum()
```

2. df.dropna(axis = 0, inplace = True)
    * axis = 0 -> row.
    * axis = 1 -> column.
    * drop column or row with null or na value and inplace is true means change original dataframe.

```python
df.dropna(axis=1, inplace=True)
print(df)
```

```python
df2.dropna(axis=0, inplace=True)
print(df2)
```
