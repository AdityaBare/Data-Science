```python
import matplotlib.pyplot as plt
```

```python
x =[1,2,3,4,5,6]
y =[44,32,56,12,90,75]
```

```python
# Plot 
# ply.polt(x,y, color='color name' , linestyle='line_style' , linewidth=value, marker='marker symbol',label='label_name' )
plt.plot(x,y, color='green' ,linestyle='--' , marker='*', label='2025 sales data')



# Set Title to graph
plt.title('Bakery sales per Month')

# x Label
plt.xlabel('Month')

# y-label
plt.ylabel("Sales")

# Small box , that show labes that pass during plot
# loc is location where upper -> left or right and bottom -> left or right
plt.legend(loc='upper left' , fontsize=12)

# grid
plt.grid(color='yellow' , linestyle=':' ) # we can also pass line style




plt.show()
```

```python

plt.plot(x,y, color='green' ,linestyle='--' , marker='o', label='2025 sales data')

plt.title('Bakery sales per Month')

plt.xlabel('Month')

plt.ylabel("Sales")

plt.grid(color='gray', linestyle=":",linewidth=1)
# Axis limits
    # To control limites or range of x-axis and y-axis
plt.xlim(1,6)

# Axis Ticks
    # Ticks are the small marks on the x-axis and y-axis that show values.

plt.xticks([1,2,3,4,5,6],['M1','M2','M3','M4','M5',"M6"])
plt.show() # Also ends they current graph
```
