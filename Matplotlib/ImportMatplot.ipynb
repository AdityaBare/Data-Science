# Import Matplotlib 

```python
import matplotlib.pyplot as plt
```

* matplotlib -> (main package)Matplotlib is the library.
* pyplot -> is a module inside Matplotlib that provides functions for creating graphs.

```python
x = [1,2,3,4,5] # X-axis values (Horizontal Values)
y = [22,20,13,23,25] # Y-axis values (Vertical Values)
```

* plt.plot(x,y)
    - plot function is used to plot line graph.
* plt.show() 
    - is used to display the graph.

```python
plt.plot(x,y)
plt.show()
```
