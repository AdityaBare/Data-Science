# Scatter Plot

- A scatter plot is used to show the relationship between two numerical variables.

* It use to find co-relation between two variables.

```python
import matplotlib.pyplot as plt

A_Score = [91,88,34,92,67,75,40,66,73]
B_Score = [79,32,90,78,88,89,90,56,77]

hr = [9,7,2,10,6,8,3,4,7]
```

```python
plt.scatter(hr, A_Score , color='blue')
plt.title("Study hour vs score")
plt.xlabel("Hour")
plt.ylabel("Score")
plt.grid(True)
plt.show()
```

```python
plt.scatter(hr, A_Score , color='blue' , label='Class A')
plt.scatter(hr, B_Score , color='Red', label='Class B')
plt.title("Comparing Two classes")
plt.xlabel("Hour")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.show()
```
