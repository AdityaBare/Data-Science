# Histogram

# Uses of a histogram:
    - Shows the distribution of data.
    - Identifies the shape of the data (e.g.normal,  skewed, uniform).
    - Reveals the spread or variability of data.
    - Detects outliers and gaps.
    - Compares different data distributions.

* Key point: Unlike a bar chart, the bars in a histogram touch each other because the data is continuous.

```python
import matplotlib.pyplot as plt
scores = [45,67,89,65,47,12,21,36,90,77,54]
```

* plt.hist(data, bins=number_of_bins , color='color', edgecolor='black)
    - bins = divide data in parts.
    - hight of bar is equal to number of student lies in the given range.

```python
plt.hist(scores, bins=5, color='green',edgecolor='white')
plt.xlabel('Score Range')
plt.ylabel("Number of students")
plt.title('Score distribution of students')
plt.show()
```

```python
data = [100,2000,50,500,2,0.8]

plt.hist(data , bins=4)
plt.show()
```
