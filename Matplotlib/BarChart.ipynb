# Bar Chart

* We use bar chart mainly to compsre values between diffrent categories.
* In simple words
    - Bar chart = Compare categories

```python
import matplotlib.pyplot as plt
product = ['A','B','C','D']
sales = [1000,800,900,1200]
```

* plt.bar(x ,height , color , width,label )


```python
plt.bar(product , sales, color='blue', label='Sales 2026', width=0.6)
plt.legend()
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales Comparsion")
plt.show()
```

* Horizontal Bar graph

```python
plt.barh(product , sales, color='blue', label='Sales 2026')
plt.legend()
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales Comparsion")
plt.show()
```
