* plt.tight_layout() 
    - Automatic adjustment
    - Always use before the plt.show() function.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot(1,2,1) # 1st row , 2nd col and 1st subplot
plt.plot(x,y)
plt.title("Line Chart")

plt.subplot(1,2,2) # 1st row , 2nd col and 2nd subplot
plt.bar(x,y)
plt.title("Bar Chart")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot(1,2,1) # 1st row , 2nd col and 1st subplot
plt.plot(x,y)
plt.title("Line Chart")

plt.subplot(1,2,2) # 1st row , 2nd col and 2nd subplot
plt.bar(x,y)
plt.title("Bar Chart")

plt.tight_layout()

fig, ax = plt.subplots(1,2, figsize = (10,5))

ax[0].plot(x,y)
ax[0].set_title("Linear Chart")
ax[1].bar(x,y)
ax[1].set_title("Bar chart")

fig.suptitle('Comparison of line and bar charts') # Main title

plt.show()
```
