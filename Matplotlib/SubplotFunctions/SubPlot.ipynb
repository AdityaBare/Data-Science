* plt.subplot(nrows, ncols, index)
    - index is 1 base not 0 base.
    - And index is of subplot.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot(1,2,1) # 1st row , 2nd col and 1st subplot
plt.plot(x,y)
plt.title("Line Chart")

plt.subplot(1,2,2) # 1st row , 2nd col and 2nd subplot
plt.bar(x,y)
plt.title("Bar Chart")
plt.show()
```

```python
plt.subplot(1,2,1) # 1st row , 2nd col and 1st subplot
plt.plot(x,y)
plt.title("Line Chart")

plt.subplot(1,2,2) # 1st row , 2nd col and 2nd subplot
plt.bar(x,y)
plt.title("Bar Chart")
plt.tight_layout()
plt.show()
```

* fig , ax = plt.subplots(nrows, ncols, figsize=(width , height))
    - When you want equal size control for each part.


```python
fig, ax = plt.subplots(1,2, figsize = (10,5))

x = [1,2,3,4]
y = [10,20,15,25]

ax[0].plot(x,y)
ax[0].set_title('Line Chart')


ax[1].bar(x,y)
ax[1].set_title("Bar Chart")

fig.suptitle("Comparision of linear and bar chart ")
```
