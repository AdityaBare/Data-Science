# Pia Chart

* Pia chart represent data as slices of circle.

```python
import matplotlib.pyplot as plt

region = ['North','South','East','West']
revenue = [3000,2000,1500,1000]
```

* plt.pia(values, labels=label_list , color=color_list , autopct='%1.1f%%')
    - Values : Data to plot.
    - Labels : Name of slices.
    - Color : Set color of slice.
    - autopct (auto percentage) : Area in percentage hold by slice.

* Max to max add 5 to 6 categories 

```python
plt.pie(revenue, labels=region, colors=['Blue','Green','Yellow','Red'],autopct='%1.1f%%')
plt.title("Revenue Contributed by region.")
plt.show()
```
