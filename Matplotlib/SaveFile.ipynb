# Saving Charts

```python
import matplotlib.pyplot as plt
x = [2,6,8,9,4]
y = [90, 89, 78,67 ,87]
```

* savefig('filename.extension', dpi=value, bbox_inches = 'tight )
    - extension -> store in format (png , pdf)
    - dpi -> dot per inch (control image resolution)
    - bbox_inches -> (tight) get fit in image or their should no extra padding.
    - Always use before plt.show()

```python
plt.plot(x,y ,color='green')
plt.scatter(x,y, color='red')
plt.title("Line Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.savefig('line_plot.png', dpi=300 , bbox_inches='tight')
plt.show()
```
